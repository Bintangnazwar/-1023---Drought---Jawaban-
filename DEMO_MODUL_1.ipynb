```python
import random

def tugas1(data_list, nim):
    # a. Tambahkan 10 elemen numerik acak (int dan float)
    for _ in range(10):
        if random.choice([True, False]):
            data_list.append(random.randint(1, 100))
        else:
            data_list.append(round(random.uniform(1, 100), 2))

    print("a. Setelah tambah 10 elemen:", data_list)

    # b. Hapus elemen pada indeks sesuai NIM ganjil/genap
    if nim % 2 == 0:
        # Jika NIM genap → hapus indeks genap
        data_list = [data_list[i] for i in range(len(data_list)) if i % 2 != 0]
    else:
        # Jika NIM ganjil → hapus indeks ganjil
        data_list = [data_list[i] for i in range(len(data_list)) if i % 2 == 0]

    print("b. Setelah hapus indeks sesuai NIM:", data_list)

    # c. Balik urutan elemen
    data_list.reverse()
    print("c. Setelah dibalik:", data_list)

    # d. Hapus elemen dengan nilai terbesar
    if data_list:
        nilai_terbesar = max(data_list)
        data_list.remove(nilai_terbesar)

    print("d. Setelah hapus nilai terbesar:", data_list)

    # e. Urutkan elemen dalam urutan menurun
    data_list.sort(reverse=True)
    print("e. Setelah diurutkan menurun:", data_list)

    return data_list


# =========================
# PROGRAM UTAMA (AGAR BISA DIRUN)
# =========================

# Input NIM dari user
nim = int(input("Masukkan NIM kamu: "))

# List awal
data_awal = [10, 20, 30, 40]

# Panggil fungsi
hasil = tugas1(data_awal, nim)

print("\nHasil akhir:", hasil)
```

```python
# Membuat tuple berisi 1 dictionary dengan 10 pasangan kunci-nilai
data_nilai = ({
    "Matematika": 85,
    "Fisika": 90,
    "Kimia": None,
    "Biologi": 78,
    "Informatika": 92,
    "Statistika": None,
    "Algoritma": 88,
    "Basis Data": 80,
    "Jaringan": None,
    "AI": 95
},)

print("Data awal:")
print(data_nilai)

# Ambil dictionary dari tuple
nilai_dict = data_nilai[0]

# a. Hapus entri yang memiliki nilai kosong (None)
nilai_bersih = {k: v for k, v in nilai_dict.items() if v is not None}

print("\n a. Setelah hapus nilai kosong:")
print(nilai_bersih)

# b. Hitung jumlah entri yang tersisa
jumlah_entri = len(nilai_bersih)
print("\n b. Jumlah entri tersisa:", jumlah_entri)

# c. Urutkan berdasarkan nilai menaik
urut_menaik = sorted(nilai_bersih.items(), key=lambda x: x[1])
print("\n c. Urut berdasarkan nilai (menaik):")
print(urut_menaik)

# d. Mata kuliah dengan nilai tertinggi
nilai_tertinggi = max(nilai_bersih, key=nilai_bersih.get)
print("\n d. Mata kuliah dengan nilai tertinggi:")
print(nilai_tertinggi, "=", nilai_bersih[nilai_tertinggi])

# e. Hitung rata-rata nilai
rata_rata = sum(nilai_bersih.values()) / len(nilai_bersih)
print("\n e. Rata-rata nilai:", rata_rata)
```

```python
# Import library
import pandas as pd

data = {
    "Merk": ["Apple", "Xiaomi", "Samsung", "Infinix", "Oppo",
             "Asus", "Lenovo", "Huawei", "Google Pixel", "Sony"],
    "Tahun Rilis": [2015, 2017, 2018, 2015, 2017,
                    2020, 2019, 2019, 2020, 2017],
    "Terjual": [150, 75, 120, 85, 75,
                120, 50, 90, 55, 50],
    "Harga (juta)": [19, 9, 10, 9, 4,
                     9, 15, 15, 15, 11],
    "Jenis Penjualan": ["Online", "Offline", "Online", "Online", "Offline",
                        "Online", "Offline", "Online", "Offline", "Offline"]
}

df = pd.DataFrame(data)

print("=== DataFrame ===")
print(df)
```

```python
# a. Statistik Deskriptif

print("\n=== Statistik Deskriptif ===")
deskriptif = df[["Tahun Rilis", "Terjual", "Harga (juta)"]].describe()
print(deskriptif)

print("\nNilai Minimum:")
print(df[["Tahun Rilis", "Terjual", "Harga (juta)"]].min())

print("\nNilai Maksimum:")
print(df[["Tahun Rilis", "Terjual", "Harga (juta)"]].max())

print("\nJumlah Total:")
print(df[["Tahun Rilis", "Terjual", "Harga (juta)"]].sum())
```

```python
# b. 3 Merek dengan Pendapatan Terbesar

df["Pendapatan"] = df["Terjual"] * df["Harga (juta)"]

top3 = df.sort_values(by="Pendapatan", ascending=False).head(3)

print("\n=== 3 Merek dengan Pendapatan Terbesar ===")
print(top3[["Merk", "Pendapatan"]])


```

```python
# c. Kelompok berdasarkan Jenis Penjualan

grouped = df.groupby("Jenis Penjualan")["Merk"].apply(list)

print("\n=== Kelompok Berdasarkan Jenis Penjualan ===")
print(grouped)
```

```python
# d. Kategori Harga (Murah, Sedang, Mahal)

q1 = df["Harga (juta)"].quantile(0.25)
q3 = df["Harga (juta)"].quantile(0.75)

def kategori_harga(harga):
    if harga < q1:
        return "Murah"
    elif harga > q3:
        return "Mahal"
    else:
        return "Sedang"

df["Kategori Harga"] = df["Harga (juta)"].apply(kategori_harga)

print("\n=== DataFrame dengan Kategori Harga ===")
print(df[["Merk", "Harga (juta)", "Kategori Harga"]])
```
